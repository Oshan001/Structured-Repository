```python
data = {
    "Name": ["Ram", "Shyam", "Hari", "Sita"],
    "Marks": [56, np.nan, 80, 54]
}

df = pd.DataFrame(data)

print(df.isna().sum())
```

```python
mean_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(mean_marks)
print(df)
```
